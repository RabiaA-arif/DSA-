```python
# Two sum array using brute force
import array
numbers:array = [2,7,11,15]
target:int = 9
for a in range(len(numbers)):
  for b in range(len(numbers)):
    my_sum = numbers[a] + numbers[b]
    if my_sum == target:
      my_index=[a,b]
      print(my_index)
      break
```

```python
# password match
import string
from itertools import product

def brute_force_password_match(password,max_length):
  char = string.ascii_lowercase
  attempts = 0
  for lenght in range(1,max_length + 1):
    for attempt in product(char,repeat=lenght):
      attempts +=1
      candidate = "".join(attempt)
      if candidate == password:
        return candidate , attempts

      return None,attempts

password:str = "yes65432!@"
# print(len(password))
max_len=len(password)
match ,total_attempts = brute_force_password_match(password,max_len)

if match:
  print(f"password is {match} after {total_attempts} : attemps")

else:
  print(f"password is not found within {total_attempts} attempts")

```

## LinkedList Implementation

```python
class Node:
  def __init__(self,value):
    self.value = value
    self.next = None
```

```python
class LinkedList():
  def __init__(self , head=None):
    self.head = head
```

```python
llist = LinkedList()
llist.head = Node(1)
second = Node(2)
third = Node(3)

llist.head.next = second
second.next = third

node = llist.head
while node:
  print(node.value)
  node = node.next
```
