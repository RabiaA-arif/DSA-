```python
# password match
import string
from itertools import product

def brute_force_password_match(password,max_length):
  char = string.ascii_lowercase
  attempts = 0
  for lenght in range(1,max_length + 1):
    for attempt in product(char,repeat=lenght):
      attempts +=1
      candidate = "".join(attempt)
      if candidate == password:
        return candidate , attempts

      return None,attempts

password:str = "yes65432!@"
# print(len(password))
max_len=len(password)
match ,total_attempts = brute_force_password_match(password,max_len)

if match:
  print(f"password is {match} after {total_attempts} : attemps")

else:
  print(f"password is not found within {total_attempts} attempts")

```

### sum of two LinkedList

```python
class Node:
  def __init__(self,data):
    self.data = data
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None

  def push(self,new_data):
    new_node = Node(new_data)
    new_node.next = self.head
    self.head = new_node

  def add_two_lists(self,first,second):
    prev = None
    temp = None
    carry = 0

    while(first is not None or second is not None):
      first_data = 0 if first is None else first.data
      second_data = 0 if second is None else second.data
      sum = carry + first_data + second_data

      carry = 1 if sum>= 10 else 0

      sum = sum if sum < 10 else sum % 10

      temp = Node(sum)

      if self.head is None:
        self.head = temp
      else:
        prev.next = temp


      prev = temp


      if first is not None:
        first = first.next
      if second is not None:
        second = second.next

      if carry > 0:
        temp.next = Node(carry)

  def printList(self):
        temp = self.head
        while(temp):
          print(f"{temp.data}->",end="")
          # print(temp.data)
          temp = temp.next


first = LinkedList()
second = LinkedList()

first.push(1)
first.push(2)
first.push(3)
# first.push(6)
# first.push(8)
print("\nFirst LinkedList")
first.printList()

second.push(9)
second.push(9)
second.push(9)
print("\nSecond LinkedList")
second.printList()

res = LinkedList()
res.add_two_lists(first.head, second.head)
print("\nAddition of Two LinkedList")
res.printList()
```

## median of two numbers

```python
import statistics

def array_median(num1: list[int], num2: list[int]) -> float:
  merged = num1 + num2
  merged.sort()
  mediaan = statistics.median(merged)
  print(f"Merged Of Two Array:{merged}")
  print(f"Median of:{mediaan}")

obj = array_median([2,3],[1,5])
obj1 = array_median([3,4,5],[1])
```

```python
class Solution:
  def find_median_sorted_array(self, num1: list[int], num2: list[int]):
    merged = []
    i = 0
    j = 0

    while i < len(num1) and j < len(num2):
      if num1[i] < num2[j]:
        merged.append(num1[i])
        i+=1
      else:
        merged.append(num2[j])
        j+=1
    print(f"Merged Array:{merged}")

    while i < len(num1):
      merged.append(num1[i])
      i+=1

    while j < len(num2):
      merged.append(num2[j])
      j+=1
    print(f"Merged Array:{merged}")

    mid = len(merged) // 2
    if len(merged) % 2 ==0:
      return (merged[mid -1] + merged[mid] /2)

    else:
      return merged[mid]


obj = Solution()
obj.find_median_sorted_array([2,3,1],[7,3])
```

```python
class Solution:
    def findMedianSortedArrays(self, nums1: list[int], nums2: list[int])-> float:
        merged = []
        i, j = 0, 0

        while i < len(nums1) and j < len(nums2):
            if nums1[i] < nums2[j]:
                merged.append(nums1[i])
                i += 1
            else:
                merged.append(nums2[j])
                j += 1

        while i < len(nums1):
            merged.append(nums1[i])
            i += 1

        while j < len(nums2):
            merged.append(nums2[j])
            j += 1

        mid = len(merged) // 2
        if len(merged) % 2 == 0:
            return (merged[mid-1] + merged[mid]) / 2
        else:
            return merged[mid]

```
