## Brute Force Algorithm
`most simplest technique to iterate every possible conditions`


```python
# bruteforce technique

pin:int = 5654 # per digit possibility is 10  and number can be repeated so 10*10*10*10=10 powr 4
counter=0
for i in range(0,10000):
  # print(f"{i}:iteration number")
  if i == 1111:
    print("pin is found")
    break
  counter+=1
print(counter)


```

```python
# Two sum array using brute force
import array
numbers:array = [2,7,11,15]
target:int = 9
for a in range(len(numbers)):
  for b in range(len(numbers)):
    my_sum = numbers[a] + numbers[b]
    if my_sum == target:
      my_index=[a,b]
      print(my_index)
      break
```
