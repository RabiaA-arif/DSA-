# Basic Operation of data Structure

Traversal ,Access,Insertion,Delete,Search,Update

Two method for traversal : linear using loop and non linear using recurssion

```python
# Array traversal using linear method
def traverse(arr: list[int]):
  for i in range(len(arr)):
    print(arr[i])
```

```python
# Linklist traversal using loop and recurssion
class LinkedList:
  def __init__(self,value):
    self.value = value
    self.next = None

def traverse(head: LinkedList)-> None:
  p = head
  while p is not None:
    p = p.next

def traverse(head: LinkedList) ->None:
  traverse(head.next) # recursively access head.val
```

```python
# Binary tree traversal ,non linear recursive method
class TreeNode:
  def __init__(self, val=0, left=None, right=None):
    self.val = val
    self.left = left
    self.right = right
def traverse(root: TreeNode):
  traverse(root.left)
  traverse(root.right)
```

```python
# this code is not quadratic time complexity

n = 8
m = 7
count = 0
for i in range(n):
  for j in range(i<m):  # Boolean trick :range(i < m) acts identically to range(1) because the inequality is never broken during this execution
    print(f"i value:{i}")
    print(f"j value:{j}")

    count+=1
print(count)
```

The unique mechanism driving this code is how Python treats boolean values inside a range() function. Because booleans are a subclass of integers in Python, True evaluates numerically to 1 and False evaluates to 0. Since i (0 to 4) is always strictly less than m (7), the condition i < m always returns True. This transforms the inner loop statement into for j in range(1):, which runs exactly once per outer loop iteration with j = 0.

## *Hierarchy*

This is the hierachy from least dominant to most dominant

Constant O(1)

Logarithmic O(log n)

Linear O(n)

Linearithmic O(n logn)

Quadratic O(n²)

Cubic O(n³)

Exponential O(2^n)

Factorial O(n!)

For example:

O(1) + O(n) + O(n) = O(n)

O(1) + O (logn) = O(logn)

O(logn) + O(n³) = O(n³)

O(n)+O(n) = O(n)

O(n logn) + O(1) = O(n logn)
